```python
# Prior Quality Check
import cv2, os
import matplotlib.pyplot as plt

PRIOR_DIR = 'gerber_priors'
files = sorted([f for f in os.listdir(PRIOR_DIR) if f.endswith('.png') and 'overview' not in f])
fig, axes = plt.subplots(3, 5, figsize=(20,12))
for ax, fname in zip(axes.flatten(), files):
    img = cv2.cvtColor(cv2.imread(f'{PRIOR_DIR}/{fname}'), cv2.COLOR_BGR2RGB)
    ax.imshow(img)
    ax.set_title(fname.replace('_prior.png',''), fontsize=9)
    ax.axis('off')
plt.tight_layout()
plt.show()
```
