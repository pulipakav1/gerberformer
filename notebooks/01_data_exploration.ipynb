```python
# Data Exploration
import os
from pathlib import Path

DATA = 'pcb-defect-dataset'
for split in ['train','val','test']:
    imgs = list(Path(DATA,split,'images').glob('*.jpg'))
    print(f'{split}: {len(imgs)} images')
```

```python
# Board family distribution
from collections import Counter
families = Counter()
for img in Path(DATA,'train','images').glob('*.jpg'):
    parts = img.stem.split('_')
    families[f'{parts[0]}_{parts[1]}'] += 1
for fam, count in sorted(families.items()):
    print(f'{fam:<20} {count}')
```
